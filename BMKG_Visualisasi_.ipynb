## Pendahuluan

Pemilihan teknik visualisasi dalam dashboard analisis data meteorologi bukan sekadar keputusan estetika, melainkan menyangkut **efektivitas komunikasi ilmiah**. Setiap jenis grafik memiliki karakteristik kognitif yang berbeda dalam menyampaikan pola, distribusi, dan perubahan temporal kepada pengguna. Notebook ini mendokumentasikan dasar pemilihan setiap teknik visualisasi yang digunakan dalam dashboard, disertai justifikasi dari literatur ilmiah bidang visualisasi data dan meteorologi terapan.

Kerangka evaluasi yang digunakan merujuk pada prinsip *Semiology of Graphics* (Bertin, 1983) dan teori *Grammar of Graphics* (Wilkinson, 2005), yang kemudian diadaptasi oleh Wickham (2010) menjadi panduan umum pemilihan chart berdasarkan jenis data dan tujuan analisis.

## 1. Time Series dengan Confidence Band — Suhu Udara Harian

**Teknik yang digunakan:** *Line chart* dengan rentang bayangan (*shaded band*) antara suhu maksimum dan minimum.

**Alasan pemilihan:**

Suhu udara adalah variabel kontinu yang berubah secara berurutan dalam waktu, sehingga representasi terbaik adalah grafik garis (*line chart*) yang mempertahankan kontinuitas temporal. Acharya et al. (2020) menyatakan bahwa *line charts* merupakan representasi paling efektif untuk data deret waktu meteorologi karena memungkinkan identifikasi tren, siklus, dan anomali secara intuitif.

Penambahan *shaded band* (area bayangan antara Tmax dan Tmin) mengikuti rekomendasi Hintze & Nelson (1998) yang memperkenalkan konsep *violin plots* dan *band charts* sebagai cara memvisualisasikan variabilitas distribusi. Dalam konteks klimatologi, rentang Tmax–Tmin merupakan indikator amplitudo termal harian (*diurnal temperature range*) yang secara ilmiah bermakna — nilai yang tinggi mengindikasikan kondisi benua (*continental*) dan nilai rendah mengindikasikan pengaruh maritim (Davy et al., 2017).

**Skala warna:**  
Merah untuk Tmax dan biru untuk Tmin mengikuti konvensi universal dalam meteorologi dan kartografi iklim (Brewer, 1994) serta panduan visualisasi IPCC Assessment Reports.

> **Referensi:**  
> Acharya, T. D., et al. (2020). Spatial visualization and analysis of climate data. *Remote Sensing*, 12(4), 588. https://doi.org/10.3390/rs12040588  
> Davy, R., et al. (2017). Diurnal asymmetry to the observed global warming. *International Journal of Climatology*, 37(1), 79–93. https://doi.org/10.1002/joc.4688 *(Scopus Q1)*  
> Hintze, J. L., & Nelson, R. D. (1998). Violin plots: A box plot-density trace synergism. *The American Statistician*, 52(2), 181–184.

## 2. Heatmap Per Jam — Pola Temporal Dua Dimensi

**Teknik yang digunakan:** *Heatmap* dengan sumbu X = jam UTC dan sumbu Y = tanggal.

**Alasan pemilihan:**

Heatmap dipilih untuk memvisualisasikan data dengan dua dimensi kategoris (hari dan jam) secara bersamaan. Teknik ini unggul dalam mengungkap **pola siklus diurnal** — variasi suhu, kelembaban, atau tekanan yang berulang setiap 24 jam — yang merupakan karakteristik penting dalam klimatologi tropis.

Menurut Wilkinson & Friendly (2009), heatmap merupakan teknik yang tepat untuk matriks data berdimensi dua di mana warna menjadi kanal visual (*visual channel*) ketiga selain posisi horizontal dan vertikal. Shneiderman (1992) juga menunjukkan bahwa *treemaps* dan heatmap secara konsisten lebih efisien daripada representasi tabular dalam mengidentifikasi kluster dan outlier pada data multi-dimensi.

Dalam meteorologi, visualisasi heatmap jam×tanggal digunakan secara luas untuk analisis *urban heat island* (Imhoff et al., 2010) dan studi pola konveksi tropis.

**Skala warna (*colorscale*):**  
Skala `RdYlBu_r` (merah-kuning-biru terbalik) dipilih untuk suhu karena memiliki sifat *diverging* yang jelas dan telah diuji aksesibilitasnya terhadap pengguna dengan buta warna parsial (Kovesi, 2015).

> **Referensi:**  
> Imhoff, M. L., et al. (2010). Remote sensing of the urban heat island effect across biomes in the continental USA. *Remote Sensing of Environment*, 114(3), 504–513. https://doi.org/10.1016/j.rse.2009.10.008 *(Scopus Q1)*  
> Kovesi, P. (2015). Good colour maps: How to design them. *arXiv*:1509.03700. https://arxiv.org/abs/1509.03700  
> Wilkinson, L., & Friendly, M. (2009). The history of the cluster heat map. *The American Statistician*, 63(2), 179–184.

## 3. Bar Chart — Curah Hujan Harian dan Klimatologi Bulanan

**Teknik yang digunakan:** *Bar chart* (batang vertikal) dengan gradasi warna berdasarkan intensitas.

**Alasan pemilihan:**

Curah hujan adalah variabel diskret berbasis kejadian (*event-based*) yang diukur dalam satuan akumulasi harian. Cleveland & McGill (1984) dalam studi klasik mereka tentang persepsi visual grafik menunjukkan bahwa manusia paling akurat membandingkan panjang batang dibandingkan area, sudut, atau kemiringan — menjadikan *bar chart* sebagai pilihan optimal untuk perbandingan nilai diskret.

Untuk visualisasi klimatologi bulanan, *bar chart* dipilih karena bulan adalah variabel kategoris ordinal. Referensi operasional BMKG dalam *Buku Panduan Klimatologi* menggunakan representasi batang untuk diagram ombrothermic (Gaussen, 1952), yang memperlihatkan hubungan curah hujan dengan suhu secara visual.

**Gradasi warna intensitas:**  
Intensitas curah hujan dikodekan dengan gradasi biru (terang = ringan, gelap = lebat) mengikuti klasifikasi BMKG: ringan (<5mm), sedang (5–20mm), lebat (20–50mm), sangat lebat (>50mm). Pendekatan ini konsisten dengan panduan *WMO Guidelines on the Calculation of Climate Normals* (WMO-No. 1203, 2017).

> **Referensi:**  
> Cleveland, W. S., & McGill, R. (1984). Graphical perception: Theory, experimentation, and application to the development of graphical methods. *Journal of the American Statistical Association*, 79(387), 531–554. *(Scopus Q1)*  
> WMO. (2017). *WMO guidelines on the calculation of climate normals* (WMO-No. 1203). World Meteorological Organization.

## 4. Wind Rose — Distribusi Arah dan Kecepatan Angin

**Teknik yang digunakan:** *Polar bar chart* (wind rose) 16 sektor × 5 kelas kecepatan.

**Alasan pemilihan:**

Arah angin adalah variabel **sirkular** yang tidak dapat divisualisasikan secara efektif menggunakan grafik kartesian konvensional. Wind rose merupakan standar de facto dalam meteorologi dan klimatologi untuk menyampaikan distribusi gabungan arah dan kecepatan angin secara simultan dalam satu representasi yang intuitif.

Mardia & Jupp (2000) menegaskan bahwa statistik sirkular memerlukan representasi polar karena konsep rata-rata, varians, dan distribusi pada data angular berbeda secara fundamental dari statistik linear. Wind rose memungkinkan pembaca langsung memahami **arah dominan angin** (panjang bar terpanjang), **frekuensi** (panjang relatif), dan **distribusi kecepatan** (komposisi warna) dalam satu pandangan.

Pembagian 16 sektor (22,5° per sektor) mengikuti resolusi standar laporan SYNOP (WMO-No. 306) dan memberikan presisi yang memadai untuk analisis angin lokal tanpa mengorbankan keterbacaan (Ackermann, 1983).

Pembagian kelas kecepatan (0–2, 2–5, 5–10, 10–20, >20 m/s) mengacu pada skala **Beaufort** yang dimodifikasi untuk penggunaan meteorologi permukaan, sebagaimana digunakan dalam publikasi *WMO Technical Regulations* (WMO-No. 49).

> **Referensi:**  
> Mardia, K. V., & Jupp, P. E. (2000). *Directional statistics*. John Wiley & Sons. https://doi.org/10.1002/9780470316979 *(Scopus-indexed)*  
> WMO. (2019). *Manual on codes* (WMO-No. 306). World Meteorological Organization.  
> Ackermann, G. R. (1983). Means and standard deviations of horizontal wind components. *Journal of Climate and Applied Meteorology*, 22(5), 959–961.

## 5. Dual-Axis Chart — Tutupan Awan dan Visibilitas

**Teknik yang digunakan:** Grafik garis dua sumbu Y (*dual-axis line chart*).

**Alasan pemilihan:**

Tutupan awan (oktas, 0–9) dan visibilitas (km, 0–80) memiliki hubungan fisis yang erat — awan tebal umumnya berkorelasi dengan penurunan visibilitas — namun memiliki satuan dan skala yang berbeda. Dual-axis chart dipilih untuk memperlihatkan kovariasi dua variabel ini dalam satu panel tanpa kehilangan informasi skala.

Meskipun beberapa sumber (Few, 2008) memperingatkan potensi misleading dari dual-axis chart, Isenberg et al. (2011) menunjukkan bahwa penggunaan dual-axis dibenarkan secara ilmiah ketika dua variabel memiliki **hubungan kausalitas atau korelasi yang mapan secara teoritis** dan tujuan visualisasi adalah memperlihatkan pola kovariasi, bukan perbandingan nilai absolut — yang persis merupakan kasus tutupan awan dan visibilitas dalam meteorologi.

> **Referensi:**  
> Isenberg, P., et al. (2011). A study on dual-scale data charts. *IEEE Transactions on Visualization and Computer Graphics*, 17(12), 2316–2325. https://doi.org/10.1109/TVCG.2011.160 *(Scopus Q1)*  
> Few, S. (2008). Dual-scaled axes in graphs: Are they ever the best solution? *Visual Business Intelligence Newsletter*.

## 6. Klimatologi Bulanan — Line Chart dengan Shaded Band untuk Suhu

**Teknik yang digunakan:** Kombinasi *shaded band* + *multi-line chart* untuk suhu; *bar chart* untuk curah hujan dan penyinaran; *area chart* untuk kelembaban.

**Alasan pemilihan:**

Visualisasi klimatologi bulanan memerlukan teknik yang berbeda untuk setiap variabel karena sifat statistik mereka berbeda:

**Suhu** memiliki variabilitas antar-bulan yang kecil di daerah tropis (sering <1°C) namun bermakna secara ilmiah. Menggunakan *bar chart* dengan sumbu Y dimulai dari nol menyembunyikan variasi ini — sebuah masalah yang dikenal sebagai *truncated axis problem* (Pandey et al., 2015). Solusinya adalah menggunakan *line chart* dengan sumbu Y yang dimulai dari nilai minimum terdekat, disertai *shaded band* Tmax–Tmin untuk menunjukkan rentang termal bulanan.

**Curah hujan** bersifat diskret dan non-negatif, sehingga *bar chart* dengan sumbu Y dimulai dari nol adalah representasi yang tepat (Cleveland & McGill, 1984).

**Kelembaban** memiliki karakteristik time-series yang kontinu, sehingga *area chart* memberikan kesan kuantitas kumulatif yang sesuai dengan interpretasi fisiknya.

> **Referensi:**  
> Pandey, A. V., et al. (2015). How deceptive are deceptive visualizations? An empirical analysis of common distortion techniques. *ACM CHI Conference on Human Factors in Computing Systems*, 1469–1478. https://doi.org/10.1145/2702123.2702608 *(Scopus Q1)*  
> Cleveland, W. S., & McGill, R. (1984). Graphical perception. *JASA*, 79(387), 531–554.

## 7. Glassmorphism dan Animated Weather Background

**Teknik yang digunakan:** CSS *glassmorphism* untuk panel data; CSS-animated background berdasarkan kode cuaca WW.

**Alasan pemilihan:**

Desain interace bukan aspek dekoratif semata karena akan mempengaruhi performa kognitif pengguna dalam membaca dan menginterpretasi data. Penerapan visual hierarchy yang jelas, termasuk pemisahan visual antara panel *kondisi terkini* dan panel *analisis periode*, mengikuti prinsip *Gestalt* yang diaplikasikan dalam desain informasi: kedekatan (*proximity*) dan pelingkupan (*enclosure*) membantu pengguna mengelompokkan informasi yang berhubungan (Ware, 2012).

Efek *glassmorphism* (transparansi + blur latar) dipilih karena memberikan kedalaman hierarki visual tanpa mengorbankan keterbacaan teks — berbeda dengan opasitas penuh yang menciptakan batas keras yang mengganggu aliran visual (Nielsen, 2000).

Latar animasi cuaca (hujan, awan, kabut) berbasis CSS murni merupakan implementasi *skeuomorphic design* yang minimal — memberikan konteks situasional kepada pengguna tentang kondisi cuaca terkini tanpa mendistraksi dari konten analitis utama (Sklar & Wasserman, 2012).

> **Referensi:**  
> Ware, C. (2012). *Information visualization: Perception for design* (3rd ed.). Morgan Kaufmann. *(Scopus-indexed)*  
> Nielsen, J. (2000). *Designing web usability*. New Riders Press.  
> Sklar, J., & Wasserman, N. (2012). *Principles of web design* (5th ed.). Cengage Learning.
